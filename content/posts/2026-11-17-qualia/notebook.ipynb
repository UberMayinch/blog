## Qualitative computing, and the analog second

"Qualia per second" still tries to make it quantitative. 10 bits/s is a rate *after* you've decided what the symbols are. The paper's binning critique bites here: bits presuppose an alphabet, and experience doesn't come pre-tokenized.

**My bet:** the native rate is something like **40 Hz of Buddhist qualia**: discrete "mind moments" (the Abhidhamma *citta* stream) roughly at gamma frequency. Each moment isn't a bit, it's a whole field configuration. And the brain is constantly running something like a **Fractional Fourier Transform** (QRI idea) on that field: rotating between "where" (space/time domain) and "what" (frequency domain) by a continuous angle, to keep re-caching the scene. The 10 bits/s is just what leaks out the motor bottleneck; the cache itself is enormous and mostly gets *rotated*, not *transmitted*.

### What counts as work on qualitative computing?

Nobody has it, but there are threads:
- **Qualitative reasoning / qualitative physics** (Forbus, Kuipers' QSIM, de Kleer): computing with signs, orderings and landmarks instead of numbers. "Rising, below the boil point" beats 87.3 °C for a lot of reasoning. Qualitative in the symbolic sense, not the phenomenal one.
- **Integrated Information Theory** (Tononi): tries to read experience off a system's causal structure. Still collapses it to a scalar (Φ), which is the same quantitative move.
- **QRI's Symmetry Theory of Valence**: valence as symmetry of the experiential field. The interesting part is that the *object* is a geometry, not a count.
- **Holographic / hyperdimensional computing** (Pribram's holonomic brain, Kanerva's vector symbolic architectures): information stored distributed in phase/frequency, where binding is a transform instead of an address. Closest to the FrFT picture.

### How do you even encode this into computers?

The problem is digital computers make you pick the alphabet first. Every encoding is a binning, which is exactly what the 10 bits/s paper does to behavior. If the 40 Hz moment is a field, the honest representation is a field evolving under a transform, not a token stream.

### What about analog computing?

This is where it gets fun:
- A **lens computes a Fourier transform** for free, at the speed of light. Propagation through quadratic-phase media (graded-index fibre, lens systems) gives you the **fractional** Fourier transform, with the angle set by propagation distance (Lohmann, Ozaktas). So the QRI picture has a literal physical substrate: optics.
- **Neuromorphic / analog VLSI** (Carver Mead onward), **physical reservoir computing**, Shannon's old **general-purpose analog computer**: let the physics *be* the dynamics instead of simulating it.
- **Coupled oscillators** (Kuramoto-style, oscillatory neural networks) natively live at a frequency and compute by phase-locking. A 40 Hz binding rhythm is a phase-locking story.

But analog doesn't escape the binning problem, it just hides it in the noise floor: a real analog system has maybe 8 to 12 effective bits of precision per channel. Which loops back to the microrhythm objection: maybe what matters isn't infinite resolution, but that the *resolution is set by the physics of the medium* rather than by an alphabet someone chose in advance. That's the qualitative part. The second isn't a container of N bits; it's 40 rotations of a field.
