# Category Theory 4: Functorial Semantics

- **functorial semantics**: a theory is a category (syntax); a model of it is a functor out of that category (semantics). Lawvere's 1963 thesis: an algebraic theory is a category $T$ with finite products, and its models are the finite-product-preserving functors $T \to \mathbf{Set}$.
- composition and homoiconicity are what enable category theory to be such a dynamic and expressive language.
  - "homoiconicity" is borrowed from programming languages (code is data, as in Lisp), so here it's a metaphor: the tools are made of the same stuff as the things they study. Categories and functors form a category $\mathbf{Cat}$; functors $C \to D$ and natural transformations form a category $[C, D]$.
- we can use functors to give the syntax of one category semantics in another: $F : T \to C$ sends every object and arrow of the syntax $T$ to its meaning in $C$, and $F(g \circ f) = F(g) \circ F(f)$ makes that meaning compositional.
- speedrun to "monad is a monoid in the category of endofunctors":
  1. a monoid in a monoidal category $(\mathcal{M}, \otimes, I)$ is an object $M$ with $\mu : M \otimes M \to M$ and $\eta : I \to M$, associative and unital.
  2. endofunctors $[C, C]$ form a monoidal category with $\otimes = $ composition and $I = \mathrm{Id}_C$ (not the cartesian product).
  3. a monoid there is $T : C \to C$ with $\mu : T \circ T \Rightarrow T$ and $\eta : \mathrm{Id}_C \Rightarrow T$: a monad.
  - the line comes from Mac Lane's *Categories for the Working Mathematician*; the "what's the problem?" and the Wadler attribution are James Iry's 2009 joke.
- e.g. semantics in neuroscience (Marr's three levels, *Vision*, 1982):
1. computation = what computation is being done, and why (the goal).
2. algorithm = what algorithm is being implemented, on what representation.
3. implementation = biological mechanism.

  Reading each level as the semantics of the one below it is my analogy, not Marr's.
